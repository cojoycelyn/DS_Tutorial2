```python
##Part 1-Google Colab and Python
#1.Create a new notebook and code the following
#1.Create a numpy array of size 2x2

#numpy is a python library that allows you to work with numbers and large datasets etc

import numpy as np

array_2x2=np.array([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
#2.Create a random array numpy array of size 3×4
#The first number is the row and second is the column
#So this is a grid of 3 rows and 4 columns
#this allows you to make an array of shape 3,4 and fill it with random numbers.Your numbers will be float

random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
#3.Create a numpy array of ones and a numpy array of zeros
ones_array = np.ones((3,3))
zeros_array=np.zeros((3,3))

print("Array of Ones:")
print(ones_array)

print("Array of zeros:")
print(zeros_array)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it
#Create a random mupy array of size 6X3

random6x3=np.random.rand(6,3)
print("Original 6x3 array:")
print(random6x3)
#Transpose it
transposed_array=random6x3.T
print("\nTransposed array(3x6):")
print(transposed_array)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product
#Define dimensions nxp and pxm
n,p,m =3,4,2

#Create two random numpy arrays
array_np=np.random.rand(n,p)
array_pm=np.random.rand(p,m)

#Compute dot product
dot_result=np.dot(array_np,array_pm)

print("Array(nxp):")
print(array_np)
print("\nArray(pxm):")
print(array_pm)
print("\nDot product result(nxm):")
print(dot_result)
```
